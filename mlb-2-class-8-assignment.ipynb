```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
data = iris['species'].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(data, labels = data.index, autopct =  '%1.2f%%')
plt.title('Species Distribution')
plt.show()
```

# 2. Count plot

```python
sns.countplot(data = tips, x = 'sex')
plt.title('Countplot on Gender Basis')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(data = tips,  x = 'total_bill', fill = True)
sns.rugplot(tips['total_bill'])
plt.show()
```

# 4. Violin plot

```python
sns.violinplot( tips['smoker'], color = 'green' )
plt.show()
```

# 5. Box plot

```python
sns.boxplot(data = tips,  x =  'total_bill')
plt.show()
```

# 6. Boxen plot

```python
sns.boxenplot(data=iris, x='sepal_length')
plt.show()
```

# 7. Histogram

```python
plt.hist(data=iris, x='sepal_length')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data=iris, x='petal_length', y='petal_width')
plt.show()
```

# 2. Line plot

```python
sns.lineplot(data=titanic, x='Age', y='Fare')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data = iris, x = 'petal_length', y = 'petal_width')

plt.show()

# note : joint plot is more insightful in multivariate analysis
```

# 4. Hexbin plot

```python

plt.hexbin(iris['petal_length'], iris['petal_width'], gridsize=25, cmap='Reds')
plt.colorbar(label='count in bin')
plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(data = iris, x='sepal_length', y='sepal_width', fill = True)
plt.xlabel("sepal Length")
plt.ylabel("sepal Width")
plt.title("2D KDE Plot between sepal Length and sepal Width")
plt.show()
```

# 6. Box plot

```python
sns.boxplot(data=tips, x="tip", y="sex")
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(data=tips, x="tip", y="sex")
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(data=tips, x="tip", y="sex")
plt.show()
```

# 9. Count plot

```python

sns.countplot(data=tips, x="day")
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
```

# 2. Pair plot

```python
sns.pairplot(iris, hue="species")
plt.show()
```

# 3. Heatmap

```python
new_iris = iris.select_dtypes(include=np.number)
sns.heatmap(new_iris.corr(), annot=True, cmap="coolwarm", fmt=".1f")
plt.title(" Heatmap of Iris")
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(tips, col = 'day',row = 'time')
g.map(sns.scatterplot,'total_bill','tip')
plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(
    titanic,
    vars=["Age", "Fare", "SibSp", "Parch"],
    hue="Sex"
)

g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)

g.add_legend()
plt.show()
```

# 6. Joint plot

```python

sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)
plt.show()
```

# 7. Boxen plot

```python
sns.boxenplot(data=tips, x="day", y="total_bill", hue="smoker")
plt.show()
```
